```python
import nltk
import random
import pickle

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Expanded dataset
qa_pairs = [
    # Python
    ("What is Python?", "Python is a programming language."),
    ("Explain Python", "Python is an easy-to-learn programming language."),
    ("Uses of Python", "Python is used in web development, AI, and data science."),
    
    # Machine Learning
    ("What is machine learning?", "Machine learning is a subset of AI."),
    ("Define machine learning", "Machine learning allows systems to learn from data."),
    ("ML applications", "ML is used in recommendation systems, prediction, and automation."),
    
    # AI
    ("What is AI?", "AI stands for Artificial Intelligence."),
    ("Define AI", "AI is the simulation of human intelligence in machines."),
    ("Applications of AI", "AI is used in chatbots, robotics, and automation."),
    
    # Data Science
    ("What is data science?", "Data science is analyzing data to get insights."),
    ("Define data science", "It involves statistics, programming, and domain knowledge."),
    
    # SQL
    ("What is SQL?", "SQL is used to manage databases."),
    ("Define SQL", "Structured Query Language is used for database operations."),
    
    # Programming Basics
    ("What is a variable?", "A variable stores data values."),
    ("What is a loop?", "A loop repeats a block of code."),
    ("What is a function?", "A function is a reusable block of code."),
    
    # DBMS
    ("What is DBMS?", "DBMS is a system to manage databases."),
    ("Types of DBMS", "Relational and Non-relational DBMS."),
    
    # OOP
    ("What is OOP?", "Object Oriented Programming is based on objects."),
    ("Concepts of OOP", "Encapsulation, Inheritance, Polymorphism."),
    
    # General Education
    ("What is computer?", "A computer is an electronic device."),
    ("What is internet?", "Internet is a global network."),
    ("What is algorithm?", "An algorithm is a step-by-step solution.")
]

# Separate questions and answers
questions = [q for q, a in qa_pairs]
answers = [a for q, a in qa_pairs]

# Vectorization
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(questions)

# Train model
model = LogisticRegression()
model.fit(X, answers)

# Save model
pickle.dump(model, open("chatbot_model.pkl", "wb"))
pickle.dump(vectorizer, open("vectorizer.pkl", "wb"))

print("Model saved successfully!")
```

```python
import pickle

# Load model
model = pickle.load(open("chatbot_model.pkl", "rb"))
vectorizer = pickle.load(open("vectorizer.pkl", "rb"))

def chatbot_response(user_input):
    input_vec = vectorizer.transform([user_input])
    return model.predict(input_vec)[0]

# Test
while True:
    user = input("You: ")
    if user.lower() == "exit":
        break
    print("Bot:", chatbot_response(user))
```
